# 02: Cars to faces
Runs car fronts through a trained autoencoder and saves the faces (used in the survey) and car|face pairs.

## Setup
Runtime > Change runtime type > **GPU**. Code lives in Google Drive at `MyDrive/carsnfaces`
(upload this folder, or sync it with Google Drive for desktop). Data, checkpoints and outputs also go to Drive,
so a disconnect loses nothing.

```
from google.colab import drive
drive.mount('/content/drive')

import os
PROJECT = '/content/drive/MyDrive/carsnfaces'   # code
os.environ['CNF_DATA'] = PROJECT + '/data'     # datasets
os.environ['CNF_RUNS'] = PROJECT + '/runs'     # checkpoints
%cd {PROJECT}
!pip install -q -e . wandb pyyaml kaggle
```

```
data, runs = os.environ['CNF_DATA'], os.environ['CNF_RUNS']
MODEL = 'faces_64'                     # or cartoon_64
CARS  = 'car_fronts_64'                # or car_rears_64 (controls)
OUT   = f'{PROJECT}/outputs/{MODEL}/{CARS}'
!python -m carsnfaces.infer --ckpt {runs}/{MODEL}/final.pt --input {data}/processed/{CARS} --out {OUT}
```

```
import glob, matplotlib.pyplot as plt
from PIL import Image
pairs = sorted(glob.glob(f'{OUT}/pairs/*.png'))[:24]
fig, axes = plt.subplots(4, 6, figsize=(15, 10))
for ax, p in zip(axes.flat, pairs):
    ax.imshow(Image.open(p), cmap='gray'); ax.set_title(os.path.basename(p)[:20], fontsize=7); ax.axis('off')
plt.tight_layout()
```
